Start by constructing the **actual SAFT-γ Mie model** for water and ethanol.
Manual construction stays light and works with core dependencies; evaluating
its EOS needs the `saft` extra. Follow the [installation guide](../installation.md)
to create the Conda environment and select its Jupyter kernel.
There is no ideal-liquid or regular-solution stand-in in this quickstart.

```python
import sys, importlib.util
from importlib.metadata import version
from thermoweave import SAFTgammaMie
model = SAFTgammaMie({'water': {'H2O':1}, 'ethanol': {'CH3':1,'CH2OH':1}}, pressure_Pa=101325.)
print('ThermoWeave version:',version('thermoweave'))
print('Python version:',sys.version.split()[0])
print('Actual model class:',type(model).__name__)
print('Components:',model.component_names)
print('Group definitions:',model.groups)
print('Model pressure [Pa]:',model.pressure_Pa)
```

```python
if importlib.util.find_spec('jax') is not None:
    T,x = 298.15,[0.8,0.2]
    rho = model.density(T,x)
    print('Liquid molar density [mol/m3]:',rho)
    print('Molar volume [cm3/mol]:',1e6/rho)
    print('ln gamma:',model.ln_activity_coefficients(T,x))
else:
    print('EOS evaluation requires the saft extra; the real model was constructed above.')
```

| Quantity | Convention |
|---|---|
| T / P | K / Pa |
| x | full mole fractions in component insertion order |
| ln γ, ln φ | dimensionless natural logarithms |
| SAFT density | mol/m³; molar volume = 1/rho in m³/mol |
| Fusion enthalpy | J/mol |
| Mixing and excess Gibbs helpers | molar energy divided by RT |
| HELD feed | species mole amounts |
| HELD `mass_fractions` | phase mass fractions; do not use as mole fractions |
